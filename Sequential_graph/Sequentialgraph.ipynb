```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class MyState(TypedDict):
    height: int
    weight: int
    BMI: float
    result: str

def calculate_bmi(state: MyState) -> MyState:
    height_m = state["height"] / 100  # Convert height to meters
    bmi = state["weight"] / (height_m ** 2)
    state["BMI"] = bmi
    return state

def check_bmi(state: MyState) -> MyState:
    bmi = state["BMI"]
    if bmi < 18.5:
        state["result"] = "You are underweight."
    elif 18.5 <= bmi < 24.9:
        state["result"] = "You have a normal weight."
    elif 25 <= bmi < 29.9:
        state["result"] = "You are overweight."
    else:
        state["result"] = "You are obese."
    return state

graph = StateGraph(MyState)
graph.add_node("calculate_bmi", calculate_bmi)
graph.add_node("check_bmi", check_bmi)
graph.set_entry_point("calculate_bmi")
graph.add_edge("calculate_bmi", "check_bmi")
graph.add_edge("check_bmi", END)

app = graph.compile()

input_state = {
    "height": 175,
    "weight": 75,
    "BMI": 0.0,
}

result = app.invoke(input_state)
print("Result:", result)
print("BMI:", result["BMI"])
print("Result:", result["result"])
```
