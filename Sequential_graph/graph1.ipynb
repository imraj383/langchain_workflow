```python
from typing import TypedDict
from langgraph.graph import StateGraph, END
```

```python
class MyState(TypedDict):
    question: str
    answer: str

def node1(state: MyState):
    return {"answer": f"You asked: {state['question']}"}

graph = StateGraph(MyState)
graph.add_node("node1", node1)
graph.set_entry_point("node1")
graph.add_edge("node1", END)

app = graph.compile()
result = app.invoke({"question": "How are you", "answer": ""})

print(result)
```

```python
class MyState2(TypedDict):
    message: str

def node2(state: MyState2):
    return {"message": f"Hey {state['message']} You are doing great by learning LangGraph"}

graph = StateGraph(MyState2)
graph.add_node("node2", node2)
graph.set_entry_point("node2")
graph.add_edge("node2", END)

app = graph.compile()
result = app.invoke({"message": "Raj"})

print(result)
```

```python
class Mystate3(TypedDict):
    name: str
    value: list
    operation: str
    result: str

def node3(state: Mystate3):
    if state['operation'] == '+':
        return {"result": f"{state['name']}, your answer is {sum(state['value'])}"}
    elif state['operation'] == '*':
        product = 1
        for num in state['value']:
            product *= num
        return {"result": f"{state['name']}, your answer is {product}"}
    else:
        return {"result": "Invalid operation"}

graph = StateGraph(Mystate3)
graph.add_node("node3", node3)
graph.set_entry_point("node3")
graph.add_edge("node3", END)

app = graph.compile()
result = app.invoke({"name": "Raj", "value": [1, 2, 3, 4], "operation": "+", "result": ""})

print(result)
```
