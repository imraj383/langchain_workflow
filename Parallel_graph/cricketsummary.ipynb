```python
from typing import TypedDict
from langgraph.graph import START, StateGraph, END

class Mystate(TypedDict):
    runss: float
    balls: float
    fours: float
    sixes: float
    strikerate: float
    runatboundary: float
    ballperboundary: float
    summary: str

def strikerate(state: Mystate) -> dict:
    if state["balls"] > 0:
        sr = (state["runss"] / state["balls"]) * 100
    else:
        sr = 0.0
    return {"strikerate": sr}

def runatboundary(state: Mystate) -> dict:
    total_boundaries = state["fours"] + state["sixes"]
    if total_boundaries > 0:
        rab = state["runss"] / total_boundaries
    else:
        rab = 0.0
    return {"runatboundary": rab}

def ballperboundary(state: Mystate) -> dict:
    total_boundaries = state["fours"] + state["sixes"]
    if total_boundaries > 0:
        bpb = state["balls"] / total_boundaries
    else:
        bpb = 0.0
    return {"ballperboundary": bpb}

def summary(state: Mystate) -> dict:
    s = (
        f"Runs: {state['runss']}, Balls: {state['balls']}, Fours: {state['fours']}, "
        f"Sixes: {state['sixes']}, Strike Rate: {state['strikerate']:.2f}, "
        f"Run at Boundary: {state['runatboundary']:.2f}, "
        f"Ball per Boundary: {state['ballperboundary']:.2f}"
    )
    return {"summary": s}

graph = StateGraph(Mystate)

graph.add_node("strikerate", strikerate)
graph.add_node("runatboundary", runatboundary)
graph.add_node("ballperboundary", ballperboundary)
graph.add_node("summary", summary)

graph.add_edge(START, "strikerate")
graph.add_edge(START, "runatboundary")
graph.add_edge(START, "ballperboundary")
graph.add_edge("strikerate", "summary")
graph.add_edge("runatboundary", "summary")
graph.add_edge("ballperboundary", "summary")
graph.add_edge("summary", END)

app = graph.compile()

input_state = {
    "runss": 50,
    "balls": 30,
    "fours": 5,
    "sixes": 2,
    "strikerate": 0.0,
    "runatboundary": 0.0,
    "ballperboundary": 0.0,
    "summary": "",
}

result = app.invoke(input_state)
print("Result:", result["summary"])
```
